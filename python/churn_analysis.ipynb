## 4. Tenure vs Churn

Analyze whether customer tenure is related to churn.


```python
# ============================================================
# EDA: TENURE vs CHURN
# Purpose: Compare customer tenure between churned and retained customers
# ============================================================

print(df.groupby("Churn")["tenure"].describe())
```

```python
# ============================================================
# EDA: TENURE vs CHURN - VISUALIZATION
# Purpose: Compare the tenure distribution of churned and retained customers
# ============================================================

df.boxplot(
    column="tenure",
    by="Churn",
    figsize=(7, 4)
)

plt.title("Tenure Distribution by Churn")
plt.suptitle("")  
plt.xlabel("Churn")
plt.ylabel("Tenure (Months)")

plt.show()
```

### Business Insight

Customers with shorter tenure tend to have a higher risk of churn compared to customers who have stayed with the company for a longer period.

This suggests that the first few months of the customer relationship are important for retention. The company should focus on improving onboarding and early customer experience to reduce early-stage churn.

## 6. Internet Service vs Churn

# Business Insight

Internet service type shows a difference in customer churn rates.

Customers using different internet service types have different levels of churn, which indicates that the type of service may be associated with customer retention.

The company should investigate service quality, pricing, and customer experience for the internet service category with the highest churn rate.

## 7. Payment Method vs Churn

Analyze whether customer churn varies across different payment methods.

```python
# ============================================================
# EDA: PAYMENT METHOD vs CHURN
# Purpose: Calculate churn rate for each payment method
# ============================================================

payment_churn_rate = pd.crosstab(
    df["PaymentMethod"],
    df["Churn"],
    normalize="index"
) * 100

print(payment_churn_rate.round(2))
```

```python
# ============================================================
# EDA: PAYMENT METHOD vs CHURN - 100% STACKED BAR CHART
# Purpose: Compare the proportion of churned and retained
#          customers across different payment methods
# ============================================================

payment_churn_pct = pd.crosstab(
    df["PaymentMethod"],
    df["Churn"],
    normalize="index"
) * 100

payment_churn_pct.plot(
    kind="bar",
    stacked=True,
    figsize=(9, 5)
)

plt.title("Churn Distribution by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Percentage of Customers")
plt.xticks(rotation=30)
plt.legend(title="Churn")

plt.show()
```

### Business Insight

Churn rates vary across different payment methods.

This suggests that payment method may be associated with customer retention. The company should investigate whether certain payment methods are linked with billing difficulties, customer preferences, or lower engagement and consider targeted retention strategies for high-churn payment segments.

## 8. Customer Services vs Churn

Analyze whether customers who use additional services have different churn rates compared to customers who do not use those services.

```python
# ============================================================
# EDA: CUSTOMER SERVICES vs CHURN
# Purpose: Calculate churn rates for different customer services
# ============================================================

service_columns = [
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies"
]

for service in service_columns:
    service_churn_rate = pd.crosstab(
        df[service],
        df["Churn"],
        normalize="index"
    ) * 100

    print("\n" + "=" * 50)
    print(service)
    print("=" * 50)
    print(service_churn_rate.round(2))
```

```python
# ============================================================
# EDA: CUSTOMER SERVICES vs CHURN - SUMMARY
# Purpose: Create a single table showing churn rate for
#          each customer service
# ============================================================

service_columns = [
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies"
]

service_churn_summary = {}

for service in service_columns:
    churn_rate = pd.crosstab(
        df[service],
        df["Churn"],
        normalize="index"
    ) * 100

    service_churn_summary[service] = churn_rate["Yes"]

service_churn_summary = pd.DataFrame(service_churn_summary)

print(service_churn_summary.round(2))
```

```python
# ============================================================
# EDA: CUSTOMER SERVICES vs CHURN - VISUALIZATION
# Purpose: Compare churn rates across different customer services
# ============================================================

service_churn_summary.T.plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Churn Rate by Customer Services")
plt.xlabel("Customer Service")
plt.ylabel("Churn Rate (%)")
plt.xticks(rotation=30)
plt.legend(title="Service Status")

plt.show()
```

### Business Insight

Churn rates differ between customers who use additional services and those who do not.

Customers without services such as Online Security and Tech Support generally show higher churn compared with customers who have these services.

This suggests that additional services may improve customer engagement and retention. The company can consider promoting useful add-on services to customers who are at higher risk of churn.
